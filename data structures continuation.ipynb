### Tuples
Tuple is similar to list but the difference is that you can't change the values once it is defined (termed as `immutability`). Due to this property it can be used to keep things that you do not want to change in your program. Example can be a country codes, zipcodes, etc...

```python
tup = (45,343,23,5,21)
tup[2]
```

### Control Flow

```python
# if statement 
iddriss = 11
age_difference = iddriss - 22
age = 22
if iddriss > age:
    print(f"You are {age_difference} older than Iddriss")
elif iddriss == age:
    print("you are the same age as iddriss")
else:
    print(f"You are {age - iddriss} year(s) younger than Iddriss")
```

```python
a = 10
while a > 20:
    print('a is: {}'.format(a))
    a = a + 1
else:
    print(f'{a} is no longer greater than 20')
```

```python
def add_nums(a,b):

  """
  Function to add two numbers given as inputs
  It will return a sum of these two numbers
  """

  sum = a+b
  
  return sum

print(add_nums(10,20))
```

```python
add_nums = lambda a,b: a+b

add_nums(10,20)
```
